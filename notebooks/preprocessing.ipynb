```python
df = pd.read_csv('../data/activity_records.csv')
df = df.drop(columns=['step', 'nameOrig', 'nameDest','isFlaggedFraud', 'oldbalanceDest', 'newbalanceDest'])
df.head()
```

```python
# Create New Features
df['trans_perc'] = df['amount'] / df['oldbalanceOrg'].replace(0, np.nan)
df['trans_perc'] = df['trans_perc'].fillna(0)

df['had_zero_orig_balance'] = (df['oldbalanceOrg'] == 0).astype(int)

df['is_cash_out'] = (df['type'] == 'CASH_OUT').astype(int)
df['is_transfer'] = (df['type'] == 'TRANSFER').astype(int)
df = df.drop(columns=['type'])

df['new_balance_orig_0'] = (df['newbalanceOrig'] == 0).astype(int)
df = df.drop(columns=['newbalanceOrig'])

df.head()
```
